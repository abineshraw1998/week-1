```python
# Week 4 – Predictive Modeling and Optimization in Logistics Systems
# Author: Abinesh

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# ---------------------------------------------------------
# 1. CREATE DUMMY LOGISTICS DATASET
# ---------------------------------------------------------

data = {
    'Distance_km': [12,25,18,30,22,15,28,35,20,26],
    'Fuel_Litres': [3.2,6.1,4.5,7.2,5.3,3.8,6.8,8.0,4.9,6.0],
    'Delivery_Time_min': [25,50,35,60,45,30,55,70,40,52]
}

df = pd.DataFrame(data)
print("Dataset:\n")
print(df)

# ---------------------------------------------------------
# 2. TRAIN LINEAR REGRESSION MODEL
# ---------------------------------------------------------

X = df[['Distance_km', 'Fuel_Litres']]
y = df['Delivery_Time_min']

model = LinearRegression()
model.fit(X, y)

# ---------------------------------------------------------
# 3. PREDICTIONS
# ---------------------------------------------------------

predictions = model.predict(X)
df['Predicted_Time'] = predictions

print("\nPredictions Added:\n")
print(df)

# ---------------------------------------------------------
# 4. MODEL EVALUATION METRICS
# ---------------------------------------------------------

mae = mean_absolute_error(y, predictions)
mse = mean_squared_error(y, predictions)
r2 = r2_score(y, predictions)

print("\nModel Evaluation:")
print(f"MAE: {mae:.2f}")
print(f"MSE: {mse:.2f}")
print(f"R² Score: {r2:.2f}")

# ---------------------------------------------------------
# 5. VISUALIZATION – ACTUAL VS PREDICTED
# ---------------------------------------------------------

plt.figure(figsize=(8,5))
plt.plot(df['Delivery_Time_min'], label='Actual Delivery Time', marker='o')
plt.plot(df['Predicted_Time'], label='Predicted Delivery Time', marker='x')
plt.title("Actual vs Predicted Delivery Time")
plt.xlabel("Route Index")
plt.ylabel("Delivery Time (min)")
plt.legend()
plt.grid(True)
plt.show()

# ---------------------------------------------------------
# 6. OPTIMIZATION INSIGHT
# ---------------------------------------------------------

print("\nOptimization Insight:")
print("Reducing distance and fuel usage improves delivery time and lowers cost. "
      "Predictive modeling helps identify efficient routes.")
```
